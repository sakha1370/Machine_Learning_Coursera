```python
import matplotlib.pyplot as plt

# Data
models = ['ANN', 'KNN', 'NMF']
rmse_values = [0.4407, 1.2972, 1.3042]

# Plotting
plt.figure(figsize=(8, 5))
plt.bar(models, rmse_values, color=['blue', 'green', 'red'])
plt.title('RMSE Comparison for ANN, KNN, and NMF')
plt.xlabel('Models')
plt.ylabel('RMSE')
plt.ylim(0, 1.5)  # Set y-axis limit for better visualization
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Add RMSE values on top of bars
for i, value in enumerate(rmse_values):
    plt.text(i, value + 0.05, f'{value:.4f}', ha='center')

plt.show()
```

```python
def find_max_subarray_indices(arr):
    max_sum = current_sum = arr[0]
    start = end = temp_start = 0

    for i, num in enumerate(arr[1:], 1):
        current_sum = num if current_sum < 0 else current_sum + num
        temp_start = i if current_sum == num else temp_start

        if current_sum > max_sum:
            max_sum = current_sum
            start, end = temp_start, i

    return start, end, max_sum

# Example usage
arr = [-7, 3, -1, 2, -4, 3]
i, j, max_sum = find_max_subarray_indices(arr)

print(f"Maximum subarray sum: {max_sum}")
print(f"Indices i and j: {i}, {j}")
print(f"Subarray: {arr[i:j+1]}")
```
